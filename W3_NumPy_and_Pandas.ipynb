#### NumPy arrays, operations, broadcasting, and Pandas ,DataFrames, Series, indexing, and data grouping.

```python
import numpy as np;
import pandas as pd;

marks = np.array([65, 72, 88, 91, 56, 79, 84, 95, 68, 73])
print(marks)
```

```python
import numpy as np                  

arr = np.array([10, 20, 30, 40, 50])
print(arr[2])          #Indexing
print(arr[1:4])        #Slicing
```

```python
arr = np.array([10, 20, 30, 40])     #Array operation
print(arr*2)
```

```python
arr = np.array([10,20,30,40,50])  #Basic statistics
print(arr.sum())
print(arr.mean())
print(arr.max())
print(arr.min())
```

```python
arr = np.array([      #Row,cols
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.shape)
```

```python
arr = np.array([10, 20, 30])  #Broadcasting

print(arr + 5)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(arr[1,1])
```
