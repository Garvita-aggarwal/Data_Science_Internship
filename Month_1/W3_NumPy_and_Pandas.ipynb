### Concatenate and Merge Dataframes(Joins)

```python
df1 = pd.DataFrame({'ID':[1,2,3,4],'Name':['Shreya','Suman','Anu','Kopal']})
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,3,4,5],'Scores':[45,34,54,65,76]})
df2
```

```python
pd.concat([df1,df2],axis=0)
```

```python
pd.concat([df1,df2],axis=1) #horizontal/Column-wise
```

```python
pd.merge(df1,df2,how='inner', on='ID')
```

```python
pd.merge(df1,df2,how='inner', left_on='ID',right_on='ID')
```
