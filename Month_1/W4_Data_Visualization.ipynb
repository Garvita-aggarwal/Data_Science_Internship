### Learned about Matplotlib (basic plots) and Seaborn (advanced plots like heatmaps, pairplots).

#### Matplotlib is a python liabrary used for creating static,animated and interactive plots.
#### Data Visualization - It is a graphical representation of Data to identify ,trends and insights.

```python

import matplotlib.pyplot as plt

#Pyplot is a collection of functions that akes atplotlib work as MatLab
```

```python
#data
x = [1,2,3,4]
y = [10,25,30,45]
plt.plot(x,y)
plt.show()
```

```python
#Customized Chart
plt.figure(figsize=(4,4)) #FigureSize
plt.plot(x,y,color='green',marker='o',linestyle='dashed',linewidth=2,markersize=12)
plt.title('Basic Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

#### Advanced Line chart -Multiple Lines & Legends

```python
#data
x =[1,2,3,4]
y1=[10,15,20,25]
y2=[12,20,24,30]
plt.plot(x,y1,label='Sales 2024') #Ploting y1 data
plt.plot(x,y2,label='Sales 2025') #Ploting y2 data
plt.title('YOY')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

#### Bar chart - Categorical Data

```python
#data
x=['A','B','C','D']
y=[12,44,36,48]
plt.bar(x,y,color='g')
plt.title("Simple bar chart")
plt.show()
```

#### Histogram - Used for Distribution Analysis(Continous Data)

```python
import random
data = [random.randint(1,10) for _ in range(100)]
plt.hist(data,bins=5)
plt.title("Histogram Example")
plt.show()
```

```python
import random
data = [random.randint(1,20) for _ in range(500)]
plt.hist(data,bins=20)
plt.title("Histogram Example")
plt.show()
```

#### Pie Chart -Used to show par-to-whole relationship/data

```python
#data
categories =['A','B','C','D']
sales = [10,24,30,44]

plt.pie(sales,labels=categories,autopct='%1.1f%%',startangle=90)
plt.title("Pie chart Example")
plt.show()
```

#### Scatter Plot - Used to find relationship b/w variables

```python
# data 
y1 = [10,20,25,30,35]
y2 = [20,30,35,45,55]

plt.scatter(y1,y2)
plt.title("Scatter Plot Example")
plt.show()
```

#### Subplots - Used to show ultiple chart in one figure

```python
#data-1 bar chart
categories =['Mon','Tues','Wed','Thur']
sales = [10,24,30,44]

#data-2 Scatter plot
y1 = [10,20,25,30,35]
y2 = [20,30,35,45,55]

plt.figure(figsize=(6,4))

#first plot - bar chart
plt.Subplot(1,2,1) #row,column,position
plt.bar(categories,sales)
plt.title('Daily Sales')
plt.xlabel('Week Days')
plt.ylabel('Sales')

#Second plot - scatter plot
plt.Subplot(1,2,2) #row,column,position
plt.scatter(y1,y2)
plt.title('User Examples')
plt.xlabel('User 1')
plt.ylabel('User 2')

plt.show()
```
